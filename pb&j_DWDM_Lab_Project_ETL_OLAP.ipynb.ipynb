```python
# Jupyter Notebook Cell 1: Load, Inspect and Clean the Online Retail II Dataset
import pandas as pd
import numpy as np

# Load dataset
df_raw = pd.read_csv("online_retail_II.csv", encoding="ISO-8859-1")

# 1. Dataset Dimensions & Basic Info
print("Raw Dimensions:", df_raw.shape)
print("\nData Types & Info:")
print(df_raw.info())

# 2. Missing Values Count
print("\nMissing Values:")
print(df_raw.isnull().sum())

# 3. Duplicate Records
print("\nDuplicate Rows:", df_raw.duplicated().sum())

# 4. Filter and Clean Dataset
# Standardize column names
df = df_raw.copy()
df.columns = df.columns.str.strip().str.replace(' ', '_')

# Remove duplicate records
df = df.drop_duplicates()

# Drop rows missing crucial identifiers (Customer ID, Invoice, Description)
df = df.dropna(subset=['Customer_ID', 'Invoice', 'Description'])

# Remove transaction cancellations (Invoice codes beginning with 'C')
df = df[~df['Invoice'].astype(str).str.startswith('C')]

# Filter out non-positive quantities and prices
df = df[(df['Quantity'] > 0) & (df['Price'] > 0)]

# Format types
df['Customer_ID'] = df['Customer_ID'].astype(int)
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
df['SalesAmount'] = df['Quantity'] * df['Price']

print("\nCleaned Dimensions:", df.shape)
print("Date Range:", df['InvoiceDate'].min(), "to", df['InvoiceDate'].max())

# Export cleaned data for use in RStudio
df.to_csv("online_retail_cleaned.csv", index=False)
print("Cleaned CSV exported as 'online_retail_cleaned.csv'")
```

```python
# Jupyter Notebook Cell 2: DW Schema Creation & Dimension/Fact Table Loading
import pandas as pd
from sqlalchemy import create_engine, text

# 1. Read the intermediate cleaned file (Read-only)
df_clean = pd.read_csv(r"C:\Users\dwara\Desktop\dwdm_project\online_retail_cleaned.csv")
df_clean["InvoiceDate"] = pd.to_datetime(df_clean["InvoiceDate"])
df_clean["full_date"] = df_clean["InvoiceDate"].dt.date

# 2. Build Dimensions with Surrogate Keys

# DIM_TIME
unique_dates = pd.Series(df_clean["full_date"].unique()).sort_values()
dim_time = pd.DataFrame({"full_date": unique_dates})
dim_time["time_key"] = range(1, len(dim_time) + 1)
dim_time["day"] = pd.to_datetime(dim_time["full_date"]).dt.day
dim_time["month"] = pd.to_datetime(dim_time["full_date"]).dt.month
dim_time["quarter"] = pd.to_datetime(dim_time["full_date"]).dt.quarter
dim_time["year"] = pd.to_datetime(dim_time["full_date"]).dt.year

# DIM_PRODUCT (Take latest distinct description per StockCode)
dim_product = df_clean[["StockCode", "Description"]].drop_duplicates(
    subset=["StockCode"], keep="last"
).copy()
dim_product.reset_index(drop=True, inplace=True)
dim_product["product_key"] = dim_product.index + 1
dim_product.rename(
    columns={"StockCode": "stock_code", "Description": "description"},
    inplace=True,
)

# DIM_CUSTOMER (One unique record per Customer_ID)
dim_customer = df_clean[["Customer_ID", "Country"]].drop_duplicates(
    subset=["Customer_ID"], keep="last"
).copy()
dim_customer.reset_index(drop=True, inplace=True)
dim_customer["customer_key"] = dim_customer.index + 1
dim_customer.rename(
    columns={"Customer_ID": "customer_id", "Country": "country"}, inplace=True
)

# 3. Build FACT_SALES by Mapping Natural Keys to Surrogate Keys
print("Building Fact Table via Surrogate Key joins...")

fact_merged = df_clean.merge(
    dim_time[["full_date", "time_key"]], on="full_date", how="inner"
)
fact_merged = fact_merged.merge(
    dim_product[["stock_code", "product_key"]],
    left_on="StockCode",
    right_on="stock_code",
    how="inner",
)
fact_merged = fact_merged.merge(
    dim_customer[["customer_id", "customer_key"]],
    left_on="Customer_ID",
    right_on="customer_id",
    how="inner",
)

fact_sales = pd.DataFrame(
    {
        "sales_key": range(1, len(fact_merged) + 1),
        "invoice_no": fact_merged["Invoice"],
        "time_key": fact_merged["time_key"],
        "product_key": fact_merged["product_key"],
        "customer_key": fact_merged["customer_key"],
        "quantity": fact_merged["Quantity"],
        "unit_price": fact_merged["Price"],
        "sales_amount": fact_merged["SalesAmount"],
    }
)

# 4. Load into SQLite Warehouse
engine = create_engine("sqlite:///retail_dw.db")

print("Writing tables to retail_dw.db...")
dim_time.to_sql("dim_time", engine, if_exists="replace", index=False)
dim_product.to_sql("dim_product", engine, if_exists="replace", index=False)
dim_customer.to_sql("dim_customer", engine, if_exists="replace", index=False)
fact_sales.to_sql("fact_sales", engine, if_exists="replace", index=False)

# Create primary/foreign key indexes for fast OLAP execution
with engine.connect() as con:
  con.execute(text("CREATE INDEX IF NOT EXISTS idx_time ON fact_sales(time_key);"))
  con.execute(
      text("CREATE INDEX IF NOT EXISTS idx_prod ON fact_sales(product_key);")
  )
  con.execute(
      text("CREATE INDEX IF NOT EXISTS idx_cust ON fact_sales(customer_key);")
  )

print("Data Warehouse build completed successfully!")
```

```python
# Jupyter Notebook Cell 3: 5 OLAP Operations
import sqlite3
import pandas as pd

conn = sqlite3.connect("retail_dw.db")

# 1. Roll-up: Aggregate daily/monthly sales up to Quarter and Year by Country
q_rollup = """
SELECT c.country, t.year, t.quarter,
       SUM(f.quantity) AS total_units,
       ROUND(SUM(f.sales_amount), 2) AS total_sales
FROM fact_sales f
JOIN dim_customer c ON f.customer_key = c.customer_key
JOIN dim_time t ON f.time_key = t.time_key
WHERE c.country IN ('United Kingdom', 'Germany', 'France')
GROUP BY c.country, t.year, t.quarter
ORDER BY c.country, t.year, t.quarter;
"""
print("=== 1. ROLL-UP (Month/Day -> Quarter & Year) ===")
df_rollup = pd.read_sql_query(q_rollup, conn)
display(df_rollup.head(8))

# 2. Drill-down: Decompose Year 2010 down to individual Months for top items
q_drilldown = """
SELECT t.year, t.month, p.description,
       SUM(f.quantity) AS total_units,
       ROUND(SUM(f.sales_amount), 2) AS total_sales
FROM fact_sales f
JOIN dim_time t ON f.time_key = t.time_key
JOIN dim_product p ON f.product_key = p.product_key
WHERE t.year = 2010 AND p.stock_code IN ('85123A', '22423')
GROUP BY t.year, t.month, p.description
ORDER BY p.description, t.month
LIMIT 8;
"""
print("=== 2. DRILL-DOWN (Year -> Month level detail) ===")
df_drilldown = pd.read_sql_query(q_drilldown, conn)
display(df_drilldown)

# 3. Slice: Restrict strictly across ONE dimension (e.g., Year = 2011)
q_slice = """
SELECT p.stock_code, p.description,
       SUM(f.quantity) AS units_sold,
       ROUND(SUM(f.sales_amount), 2) AS total_revenue
FROM fact_sales f
JOIN dim_time t ON f.time_key = t.time_key
JOIN dim_product p ON f.product_key = p.product_key
WHERE t.year = 2011
GROUP BY p.stock_code, p.description
ORDER BY total_revenue DESC
LIMIT 5;
"""
print("=== 3. SLICE (Dimension: Year = 2011) ===")
df_slice = pd.read_sql_query(q_slice, conn)
display(df_slice)

# 4. Dice: Filter across MULTIPLE dimensions (Specific Countries AND specific Quarter)
q_dice = """
SELECT c.country, t.year, t.quarter,
       COUNT(DISTINCT f.invoice_no) AS basket_count,
       ROUND(SUM(f.sales_amount), 2) AS dice_sales
FROM fact_sales f
JOIN dim_customer c ON f.customer_key = c.customer_key
JOIN dim_time t ON f.time_key = t.time_key
WHERE c.country IN ('France', 'Germany', 'Spain')
  AND t.year = 2010
  AND t.quarter IN (3, 4)
GROUP BY c.country, t.year, t.quarter;
"""
print("=== 4. DICE (Multi-dimension: Specific Countries AND Quarters) ===")
df_dice = pd.read_sql_query(q_dice, conn)
display(df_dice)

# 5. Pivot: Rotate Country dimension across Year columns
q_pivot = """
SELECT c.country,
       ROUND(SUM(CASE WHEN t.year = 2009 THEN f.sales_amount ELSE 0 END), 2) AS sales_2009,
       ROUND(SUM(CASE WHEN t.year = 2010 THEN f.sales_amount ELSE 0 END), 2) AS sales_2010,
       ROUND(SUM(CASE WHEN t.year = 2011 THEN f.sales_amount ELSE 0 END), 2) AS sales_2011
FROM fact_sales f
JOIN dim_customer c ON f.customer_key = c.customer_key
JOIN dim_time t ON f.time_key = t.time_key
WHERE c.country IN ('United Kingdom', 'Germany', 'France', 'EIRE', 'Netherlands')
GROUP BY c.country
ORDER BY sales_2010 DESC;
"""
print("=== 5. PIVOT (Countries as rows, Years as columns) ===")
df_pivot = pd.read_sql_query(q_pivot, conn)
display(df_pivot)
```
